# Introduction

This package provides examples of exploring and using resources in the
ENCORE project resources made available in NHGRI AnVIL.

We address the following concerns:

-   What ENCORE resources are available in AnVIL, and how can the
    Bioconductor AnVIL package be used to survey and retrieve them?

-   How can ENCORE resources be combined with other genomic reference
    data and experiments to foster discovery in RNA biology and its
    application to disease mechanisms?
